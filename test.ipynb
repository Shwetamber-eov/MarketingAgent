```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
import os
from dotenv import load_dotenv
load_dotenv()

def send_blog_email(
    blog_markdown: str,
    sender_app_password: str = os.getenv("GOOGLE_APP_PASSWORD"),
    sender_email: str = os.getenv("RECIPIENT_MAIL_ID"),
    recipient_email: str = os.getenv("RECIPIENT_MAIL_ID"),
    title: str = "",
    subject: str = "Your AI Generated Blog "
):
    print(sender_app_password)
    msg = MIMEMultipart("alternative")
    msg["From"] = sender_email
    msg["To"] = recipient_email
    msg["Subject"] = subject+title

    # Plain-text version
    msg.attach(MIMEText(blog_markdown, "plain", "utf-8"))
    print(msg)
    try:
        with smtplib.SMTP("smtp.gmail.com", 587) as server:
            server.starttls()
            server.login(sender_email, sender_app_password)
            server.sendmail(
                sender_email,
                recipient_email,
                msg.as_string()
            )

        return True

    except Exception as e:
        print(f"Failed to send email: {e}")
        return False

send_blog_email("heloooo")
```

```python
import base64
import os

from dotenv import load_dotenv
from google import genai


load_dotenv()


def generate_image(
    prompt_text: str,
    output_path: str = "output.png",
    aspect_ratio: str = "16:9",
    image_size: str = "1K",
):
    """
    Generate an image with Gemini 3.1 Flash Image.

    Args:
        prompt_text: Description of the image to generate.
        output_path: Where to save the generated image.
        aspect_ratio: e.g. "1:1", "16:9", "9:16".
        image_size: "1K", "2K", or "4K".
    """

    api_key = os.getenv("GOOGLE_API_KEY")

    if not api_key:
        raise RuntimeError(
            "GEMINI_API_KEY was not found. "
            "Add it to your .env file."
        )

    client = genai.Client(api_key=api_key)

    print("Generating image...")
    print(f"Prompt: {prompt_text}")

    try:
        interaction = client.interactions.create(
            model="gemini-3.1-flash-image",
            input=prompt_text,
            response_format={
                "type": "image",
                "aspect_ratio": aspect_ratio,
                "image_size": image_size,
            },
        )

        if not interaction.output_image:
            raise RuntimeError(
                "Gemini did not return an image."
            )

        image_data = base64.b64decode(
            interaction.output_image.data
        )

        with open(output_path, "wb") as f:
            f.write(image_data)

        print(f"Success! Image saved to: {output_path}")

        return output_path

    except Exception as e:
        print(f"Image generation failed: {e}")
        raise


sample_prompt = """
An isometric 3D vector diagram illustrating a
microservices network architecture.

Show multiple server nodes connected through clean,
glowing neon-blue network lines. Include API gateway,
authentication service, user service, payment service,
database nodes, message queue, and monitoring service.

Dark futuristic technology background, sophisticated
enterprise software architecture visualization,
clean geometry, minimal visual clutter, professional
modern aesthetic, crisp details, subtle depth,
high-quality 3D vector illustration.
"""


if __name__ == "__main__":
    generate_image(
        prompt_text=sample_prompt,
        output_path="architecture_concept.png",
        aspect_ratio="16:9",
        image_size="1K",
    )
```

```python
import io
import os
from huggingface_hub import InferenceClient
from PIL import Image
from dotenv import load_dotenv
load_dotenv()
# 1. Initialize the client with your Hugging Face API token
# Best practice: Set this as an environment variable, or paste your token string directly here
HF_TOKEN = os.getenv("HF_TOKEN") 
client = InferenceClient(token=HF_TOKEN)

# 2. Define the models we want to use (ordered by priority)
MODELS = {
    "flux": "black-forest-labs/FLUX.1-schnell",
    "sd35": "stabilityai/stable-diffusion-3.5-large"
}

def generate_image(prompt: str, output_path: str = "output.png"):
    """
    Tries to generate an image using FLUX.1 [schnell], 
    falling back to Stable Diffusion 3.5 if the server is busy.
    """
    # Try FLUX first
    model_name = MODELS["flux"]
    print(f"Sending prompt to: {model_name}...")
    
    try:
        # Call the Serverless API
        image_bytes = client.text_to_image(
            prompt,
            model=model_name
        )
        
        # Convert raw bytes into a viewable PIL Image and save
        image = Image.open(io.BytesIO(image_bytes))
        image.save(output_path)
        print(f" Success! Image saved to {output_path}")
        return
        
    except Exception as e:
        print(f"FLUX API encountered an error or timeout: {e}")
        print("Switching to backup model (Stable Diffusion 3.5)...")
        
    # Fallback to Stable Diffusion 3.5
    try:
        model_name = MODELS["sd35"]
        image_bytes = client.text_to_image(
            prompt,
            model=model_name
        )
        image = Image.open(io.BytesIO(image_bytes))
        image.save(output_path)
        print(f" Success with fallback model! Image saved to {output_path}")
        
    except Exception as fallback_error:
        print(f"Both models failed. Error: {fallback_error}")
        print("Tip: Check if your HF_TOKEN is valid or if you have hit the free hourly rate limit.")

# 3. Invoke the function with your prompt
my_prompt = "A sleek, high-contrast digital illustration showing a glowing, multi-node AI brain seamlessly integrating into enterprise architecture blueprints. Use a corporate tech color palette (deep navy blue, neon cyan, and crisp white). The style should be modern, minimalist B2B SaaS vector art."
generate_image(prompt=my_prompt, output_path="cyberpunk_cafe.png")
```

```python
from typing import List
from pydantic import BaseModel
from google import genai
from google.genai import types as google_types
import os
from dotenv import load_dotenv

load_dotenv()

# SEARCH_MODEL = "models/gemma-4-26b-a4b-it"
SEARCH_MODEL = "models/gemma-4-31b-it"


class SearchResult(BaseModel):
    title: str
    url: str
    summary: str


class SearchResults(BaseModel):
    results: List[SearchResult]


def _grounded_search(query: str, max_results: int = 8) -> List[dict]:
    api_key = os.getenv("GOOGLE_API_KEY")
    print("in function")
    if not api_key:
        return []

    try:
        client = genai.Client(api_key=api_key)
        print(" in try block")
        prompt = f"""
Search the web for:
{query}
Find the most relevant authoritative sources.
Return up to {max_results} sources.

For each source provide:
- title: short source title
- url: the actual source website URL
- summary: 1-2 sentence summary explaining what information the source provides
"""

        response = client.models.generate_content(
            model=SEARCH_MODEL,
            contents=prompt,
            config=google_types.GenerateContentConfig(
                tools=[
                    google_types.Tool(
                        google_search=google_types.GoogleSearch()
                    )
                ],
                response_mime_type="application/json",
                response_schema=SearchResults,
            ),
        )
        print("after calling model")
        print("response:::::::::",response)
        if not response.parsed:
            print("[warning] Could not parse structured search response")
            return []
        # print(response)
        return [
            result.model_dump()
            for result in response.parsed.results
        ]

    except Exception as exc:
        print(f"[warning] grounded search failed: {exc}")
        return []

results = _grounded_search(
    "digital product engineering services india"
)

for result in results:
    print(result)
```
